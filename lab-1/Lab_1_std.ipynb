```python

def show_images(images,titles=None):
    #This function is used to show image(s) with titles by sending an array of images and an array of associated titles.
    # images[0] will be drawn with the title titles[0] if exists
    # You aren't required to understand this function, use it as-is.
    n_ims = len(images)
    if titles is None: titles = ['(%d)' % i for i in range(1,n_ims + 1)]
    fig = plt.figure()
    n = 1
    for image,title in zip(images,titles):
        a = fig.add_subplot(1,n_ims,n)
        if image.ndim == 2: 
            plt.gray()
        plt.imshow(image)
        a.set_title(title)
        n += 1
    fig.set_size_inches(np.array(fig.get_size_inches()) * n_ims)
    plt.show()
```

```python
def gray_image(image):
    """
    Convert an image to grayscale and show
    original + grayscale side by side.
    """
    gray = rgb2gray(image)   # Convert to grayscale

    # Show original and grayscale side by side
    fig, axes = plt.subplots(1, 2, figsize=(10, 5))

    axes[0].imshow(image)
    axes[0].set_title("Original RGB Image")
    axes[0].axis("off")

    axes[1].imshow(gray, cmap="gray")
    axes[1].set_title("Grayscale Image")
    axes[1].axis("off")

    plt.show()
```

```python
pyramids = io.imread('pyramids.jpeg')
gray_image(pyramids)
```

```python
def HSV_image(image):
    """
    Convert an image to HSV and show:
    - Original RGB
    - Hue, Saturation, and Value channels
    """
    hsvImg = rgb2hsv(image)  

    # Separate channels
    H = hsvImg[:, :, 0]   # Hue
    S = hsvImg[:, :, 1]   # Saturation
    V = hsvImg[:, :, 2]   # Value

    # Show RGB + 3 channels side by side
    fig, axes = plt.subplots(1, 4, figsize=(18, 5))

    axes[0].imshow(image)
    axes[0].set_title("Original RGB")
    axes[0].axis("off")

    axes[1].imshow(H, cmap="hsv")
    axes[1].set_title("Hue Channel")
    axes[1].axis("off")

    axes[2].imshow(S, cmap="gray")
    axes[2].set_title("Saturation Channel")
    axes[2].axis("off")

    axes[3].imshow(V, cmap="gray")
    axes[3].set_title("Value Channel")
    axes[3].axis("off")

    plt.show()
```

```python

def apply_salt_pepper(image, amounts):
    """
    Convert image to grayscale, then apply
    salt & pepper noise with different amounts.
    Display results side by side.
    """
    gray = rgb2gray(image)

    fig, axes = plt.subplots(1, len(amounts) + 1, figsize=(15, 5))

    axes[0].imshow(gray, cmap="gray")
    axes[0].set_title("Original Grayscale")
    axes[0].axis("off")

    for i, amt in enumerate(amounts):
        noisy_img = random_noise(gray, mode='s&p', amount=amt)
        axes[i+1].imshow(noisy_img, cmap="gray")
        axes[i+1].set_title(f"S&P Noise\namount={amt}")
        axes[i+1].axis("off")

    plt.show()


```

```python
image = io.imread('pyramids.jpeg')
apply_salt_pepper(image, [0.05, 0.5, 0.9])
```

```python


def showHist(image, nbins=256):

    gray = rgb2gray(image)

    hist, bin_centers = histogram(gray, nbins=nbins)

    plt.figure(figsize=(6, 4))
    plt.plot(bin_centers, hist, lw=2)
    plt.title(f"Histogram with {nbins} bins")
    plt.xlabel("Intensity")
    plt.ylabel("Frequency")
    plt.show()



```

```python
img = io.imread('histogram/ex1.jpg')

showHist(img, nbins=256)
showHist(img, nbins=64)
showHist(img, nbins=8)
```

```python

def uniform_hist_image():
    """
    Create a 256x256 grayscale image
    with uniform histogram distribution.
    """
    img = np.arange(256, dtype=np.uint8)
    img = np.tile(img, (256, 1))   

    plt.figure(figsize=(5, 5))
    plt.imshow(img, cmap="gray")
    plt.title("Uniform Histogram Image (256x256)")
    plt.axis("off")
    plt.show()


    hist, bin_centers = histogram(img, nbins=256)
    plt.figure(figsize=(6, 4))
    plt.plot(bin_centers, hist, lw=2)
    plt.title("Uniform Histogram")
    plt.xlabel("Intensity")
    plt.ylabel("Frequency")
    plt.show()


```
